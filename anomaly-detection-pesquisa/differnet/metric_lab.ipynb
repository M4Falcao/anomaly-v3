```python
import zipfile
import os

def unzip_file(zip_path, extract_to='.'):
    """
    Descompacta um arquivo .zip para um diretório especificado.

    Args:
        zip_path (str): O caminho para o arquivo .zip a ser descompactado.
        extract_to (str): O diretório para onde os arquivos serão extraídos.
                          Por padrão, descompacta no diretório atual.
    """
    if not os.path.exists(zip_path):
        print(f"Erro: O arquivo .zip não foi encontrado em '{zip_path}'.")
        return

    try:
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_to)
        print(f"✅ Arquivo '{zip_path}' descompactado com sucesso para '{extract_to}'.")
    except zipfile.BadZipFile:
        print(f"Erro: '{zip_path}' não é um arquivo .zip válido ou está corrompido.")
    except Exception as e:
        print(f"Ocorreu um erro ao descompactar o arquivo: {e}")

# Exemplo de uso:
# Substitua 'caminho/para/seu/arquivo.zip' pelo caminho real do seu arquivo zip.
# Se o arquivo estiver na raiz do colab, pode ser apenas o nome do arquivo, ex: 'meu_arquivo.zip'
unzip_file('/content/anomaly-detection-dataset/insplad-seg.zip')
```

# Metric Lab: Comprehensive CAM Evaluation
This notebook contains all necessary code to calculate metrics for Class Activation Maps (CAMs).
It includes implementations for:
- **Pixel AUROC**: Anomaly localization accuracy.
- **Image AUROC**: Anomaly detection accuracy.
- **Deletion**: Faithfulness metric (removing relevant pixels decreases score).
- **Insertion**: Faithfulness metric (adding relevant pixels increases score).
- **ROAD**: Remove and Debias (noisy imputation).
- **Confidence Drop**: Simple confidence reduction.

## Setup
Load your model in the 'Model Setup' section and run the evaluation loop.

```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
import sys
import cv2
import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.notebook import tqdm
from sklearn.metrics import roc_auc_score, auc
from torchvision import models, transforms
from torch.utils.data import DataLoader, Dataset

# pytorch-grad-cam imports
try:
    from pytorch_grad_cam import GradCAM, ScoreCAM, GradCAMPlusPlus, XGradCAM, EigenCAM, EigenGradCAM
    from pytorch_grad_cam.utils.model_targets import ClassifierOutputTarget
    from pytorch_grad_cam.metrics.road import ROADMostRelevantFirst
except ImportError:
    print("Please install pytorch-grad-cam: pip install grad-cam")

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Device: {device}")
```

```python
# --- Helper Functions ---
def _get_score(output, target_category_idx):
    if output.shape[1] == 1:
        return output[0, 0].item()
    else:
        probs = F.softmax(output, dim=1)
        if target_category_idx is None:
            target_category_idx = torch.argmax(probs).item()
        return probs[0, target_category_idx].item()

# --- Metric: Pixel AUROC ---
def calculate_pixel_auroc(anomaly_map, ground_truth_mask):
    flat_anomaly = anomaly_map.flatten()
    flat_mask = ground_truth_mask.flatten()
    flat_mask = (flat_mask > 0.5).astype(int)
    if np.sum(flat_mask) == 0:
        return 0.5
    return roc_auc_score(flat_mask, flat_anomaly)

# --- Metric: Image AUROC ---
def calculate_image_auroc(anomaly_scores, true_labels):
    return roc_auc_score(true_labels, anomaly_scores)

# --- Metric: Confidence Drop ---
def confidence_drop_metric(model, input_tensor, cam_mask, target_category_idx=None):
    model.eval()
    with torch.no_grad():
        output_orig = model(input_tensor)
        score_orig = _get_score(output_orig, target_category_idx)

        cam_mask_tensor = torch.tensor(cam_mask).to(input_tensor.device).unsqueeze(0).unsqueeze(0)
        if cam_mask_tensor.shape[-2:] != input_tensor.shape[-2:]:
            cam_mask_tensor = F.interpolate(cam_mask_tensor, size=input_tensor.shape[-2:], mode='bilinear', align_corners=False)

        perturbed_input = input_tensor * (1 - cam_mask_tensor)
        output_pert = model(perturbed_input)
        score_pert = _get_score(output_pert, target_category_idx)

        confidence_drop = max(0, score_orig - score_pert)

    return confidence_drop

# --- Metric: Deletion ---
def calculate_deletion_single_image(model, input_tensor, heatmap, steps, device):
    heatmap_flat = heatmap.flatten()
    sorted_indices = np.argsort(heatmap_flat)[::-1].copy()

    with torch.no_grad():
        pred = model(input_tensor)
        initial_score = (pred ** 2).sum().item() # Assuming energy/score based

    scores = [1.0]
    if initial_score == 0: return 0.0, scores

    img_pert = input_tensor.clone()
    total_pixels = heatmap_flat.shape[0]
    step_size = max(1, total_pixels // steps)

    c, h, w = img_pert.shape[1:]

    for i in range(1, steps + 1):
        indices_to_remove = sorted_indices[:i*step_size]

        # Optimized zeroing mask
        mask = torch.ones(h * w, device=device)
        mask[indices_to_remove] = 0.0
        mask = mask.view(1, 1, h, w)
        img_pert = img_pert * mask

        with torch.no_grad():
            pred_new = model(img_pert)
            new_score = (pred_new ** 2).sum().item()

        scores.append(new_score / initial_score)

    x_axis = np.linspace(0, 1, len(scores))
    auc_val = auc(x_axis, scores)
    return auc_val, scores

# --- Metric: Insertion ---
def insertion_metric(model, input_tensor, cam_mask, target_category_idx=None, steps=20):
    model.eval()
    with torch.no_grad():
        output = model(input_tensor)
        # For insertion, we often normalize by final score, but standard is absolute score curve AUC
        # or normalized. Let's return raw AUC of probability curve.

    cam_flat = cam_mask.flatten()
    sorted_indices = np.argsort(cam_flat)[::-1].copy()

    n_pixels = len(cam_flat)
    step_size = max(1, n_pixels // steps)

    c, h, w = input_tensor.shape[1:]
    current_input = torch.zeros_like(input_tensor)
    scores = []

    # Step 0
    with torch.no_grad():
        out = model(current_input)
        score = _get_score(out, target_category_idx)
        scores.append(score)

    for i in range(1, steps + 1):
        limit = min(i * step_size, n_pixels)
        indices_to_add = sorted_indices[:limit]
        ys, xs = np.unravel_index(indices_to_add, (h, w))
        current_input[0, :, ys, xs] = input_tensor[0, :, ys, xs]

        with torch.no_grad():
            out_i = model(current_input)
            score_i = _get_score(out_i, target_category_idx)
            scores.append(score_i)

    auc_score = np.trapezoid(scores, dx=1.0/steps)
    return auc_score, scores

# --- Metric: ROAD ---
def calculate_road_metric(model, input_tensor, cam_mask, target_category_idx=None, percentiles=[10, 50]):
    if target_category_idx is None:
        with torch.no_grad():
            out = model(input_tensor)
            target_category_idx = torch.argmax(out).item()

    targets = [ClassifierOutputTarget(target_category_idx)]
    if len(cam_mask.shape) == 2:
        cam_mask_exp = cam_mask[None, :, :]
    else:
        cam_mask_exp = cam_mask

    scores = []
    for p in percentiles:
        road = ROADMostRelevantFirst(percentile=p)
        # Returns list of scores
        batch_scores = road(input_tensor, cam_mask_exp, targets, model)
        scores.append(np.mean(batch_scores))

    return np.mean(scores), scores
```

```python
def evaluate_model_cams(model, data_loader, cam_methods_dict):
    """
    Orchestrator.
    model: PyTorch model
    data_loader: Yields (images, labels, masks)
    cam_methods_dict: {'Name': CAM_Class}
    """
    results = []

    model.eval()

    # Iterate over methods
    for name, cam_cls in cam_methods_dict.items():
        print(f"Evaluating {name}...")

        # Identify target layers (Simplistic heuristic for demo)
        target_layers = None
        # Try common layers
        if hasattr(model, 'layer4'): target_layers = [model.layer4[-1]]
        elif hasattr(model, 'features'): target_layers = [model.features[-1]]
        elif hasattr(model, 'feature_extractor'): target_layers = [model.feature_extractor.features[10]]
        else:
            print("Could not automatically find target layer. Please set manually.")
            continue

        try:
            cam_extractor = cam_cls(model=model, target_layers=target_layers)
        except Exception as e:
            print(f"Error init CAM {name}: {e}")
            continue

        # Metrics storage
        metrics_store = {'deletion': [], 'insertion': [], 'road': [], 'confidence': [], 'pixel_auc': []}
        curves_store = {'deletion': [], 'insertion': [], 'road': []}

        for batch_idx, batch_data in enumerate(tqdm(data_loader, leave=False)):
            if len(batch_data) == 3:
                 images, labels, masks = batch_data
            elif len(batch_data) == 2:
                 images, labels = batch_data
                 masks = None
            else:
                 images = batch_data[0]
                 masks = None

            images = images.to(device)

            for i in range(len(images)):
                img = images[i].unsqueeze(0)

                # Generate CAM
                # Target: Max Class
                with torch.no_grad():
                    out = model(img)
                    t_idx = torch.argmax(out).item()

                targets = [ClassifierOutputTarget(t_idx)]
                cam_map = cam_extractor(input_tensor=img, targets=targets)[0, :]

                # Calculate Metrics
                # Pixel AUROC
                if masks is not None:
                    m = masks[i].cpu().numpy().squeeze()
                    metrics_store['pixel_auc'].append(calculate_pixel_auroc(cam_map, m))

                # Deletion
                d_auc, d_scores = calculate_deletion_single_image(model, img, cam_map, 20, device)
                metrics_store['deletion'].append(d_auc)
                curves_store['deletion'].append(d_scores)

                # Insertion
                i_auc, i_scores = insertion_metric(model, img, cam_map, t_idx, 20)
                metrics_store['insertion'].append(i_auc)
                curves_store['insertion'].append(i_scores)

                # ROAD
                r_score, r_scores = calculate_road_metric(model, img, cam_map, t_idx)
                metrics_store['road'].append(r_score)
                curves_store['road'].append(r_scores)

                # Confidence Drop
                metrics_store['confidence'].append(confidence_drop_metric(model, img, cam_map, t_idx))

        # Aggregate
        res = {'Method': name}
        for k, v in metrics_store.items():
            if v:
                res[k] = np.mean(v)

        # Aggregate Curves
        for k, v in curves_store.items():
            if v:
                # Ensure uniform length before mean
                try:
                    arr = np.array(v)
                    res[f'{k}_curve'] = np.mean(arr, axis=0)
                except Exception as e:
                    # If shapes differ (e.g. road percentiles mismatch), skip
                    pass

        results.append(res)

    return results
```

```python
from torch.utils.data import Dataset, DataLoader

class CombinedTestDataset(Dataset):
    """
    Dataset wrapper to return (image, label, mask) by combining
    the test dataset and ground truth dataset.
    """
    def __init__(self, test_dataset, gt_dataset):
        self.test_dataset = test_dataset
        self.gt_dataset = gt_dataset
        # Ensure datasets are aligned
        if len(self.test_dataset) != len(self.gt_dataset):
            print(f"Warning: Test set length ({len(self.test_dataset)}) differs from GT set length ({len(self.gt_dataset)})")

    def __len__(self):
        return len(self.test_dataset)

    def __getitem__(self, idx):
        # Get image and label from test set
        # Assuming test_dataset returns (image, label)
        image, label = self.test_dataset[idx]

        # Get mask from ground truth set
        # Handles cases where gt_dataset returns (mask, label) or just mask
        mask_data = self.gt_dataset[idx]
        if isinstance(mask_data, tuple):
            mask = mask_data[0]
        else:
            mask = mask_data

        return image, label, mask

def create_val_gt_dataloader(test_set, ground_truth_set, batch_size=1):
    """
    Creates a DataLoader that yields (image, label, mask).
    """
    combined_dataset = CombinedTestDataset(test_set, ground_truth_set)
    loader = DataLoader(combined_dataset, batch_size=batch_size, shuffle=False)
    return loader

# Instantiate the loader
val_gt_loader = create_val_gt_dataloader(test_set, ground_truth_set)
print(len(val_gt_loader))
```

```python
# --- Placeholder Setup ---
print("Setting up placeholder experiment...")

# 1. Model: ResNet18 (DifferNet in this case)
model = model_differnet
model.to(device)
model.eval()

loader = val_gt_loader

# Wrapper to fix 5D input shape (Batch, Crops, C, H, W) -> (Batch*Crops, C, H, W)
# AND to ensure masks are 1-channel
def flat_batch_loader(dataloader):
    for batch in dataloader:
        if len(batch) == 3:
            images, labels, masks = batch

            # --- Fix for 3-channel masks ---
            # Ensure masks are 1-channel (Grayscale).
            # If [B, 3, H, W] or [B, N, 3, H, W], take the first channel.
            if masks.shape[-3] == 3:
                if masks.dim() == 4:
                    masks = masks[:, 0:1, :, :]
                elif masks.dim() == 5:
                    masks = masks[:, :, 0:1, :, :]

            # Check for 5D tensor: [B, Crops, C, H, W]
            if images.dim() == 5:
                B, N, C, H, W = images.shape
                # Flatten B and N into the batch dimension
                images = images.view(-1, C, H, W)

                # Expand labels and masks to match the new batch size
                if labels.dim() == 1:
                    labels = labels.repeat_interleave(N)

                # Handle masks (Assuming masks match images or are per-image)
                # If masks are [B, H, W] -> [B*N, H, W]
                if masks.dim() == 3:
                    masks = masks.repeat_interleave(N, dim=0)
                elif masks.dim() == 4 and masks.shape[1] == 1: # [B, 1, H, W]
                     masks = masks.repeat_interleave(N, dim=0)
                elif masks.dim() == 4 and masks.shape[1] == N: # [B, N, H, W]
                     masks = masks.view(-1, H, W)
                elif masks.dim() == 5: # [B, N, 1, H, W]
                     masks = masks.view(-1, H, W)

            yield images, labels, masks
        else:
            yield batch

# 3. CAM Methods to Test
cams = {
    'GradCAM': GradCAM,
    'GradCAM++': GradCAMPlusPlus,
    'XGradCAM': XGradCAM,
    # 'EigenCAM': EigenCAM,
}

# 4. Run with the wrapper
print("Starting evaluation (with shape correction)...")
# try:
print(loader.batch_sampler)
results = evaluate_model_cams(model, flat_batch_loader(loader), cams)
# except Exception as e:
#     print(f"An error occurred during evaluation: {e}")
#     import traceback
#     traceback.print_exc()
#     results = []
```
